# Outliers

```python
len(num_features)
```

```python
def IQR(feature):
    global data
    
    Q1, Q3 = data[feature].quantile(0.25), data[feature].quantile(0.75)
    IQR = Q3 - Q1
    min_v = Q1 - 1.5 * IQR
    max_v = Q3 + 1.5 * IQR
    
    outlier_min =  data[data[feature] < min_v][feature].values
    outlier_max =  data[data[feature] > max_v][feature].values

    data[feature] = data[feature].replace(outlier_min, min_v)
    data[feature] = data[feature].replace(outlier_max, max_v)
```

```python

for feature in num_features:
    plt.figure(figsize=(20,2))

    plt.subplot(1, 2 , 1)
    sns.boxplot(data=data, x=feature)
    plt.xlabel(f'{feature}, Before IQR')
    plt.grid(True)
    
    IQR(feature)
    
    plt.subplot(1, 2 , 2)
    sns.boxplot(data=data, x=feature)
    plt.xlabel(f'{feature}, After IQR')
    plt.grid(True)
    
    plt.show()
    
```
